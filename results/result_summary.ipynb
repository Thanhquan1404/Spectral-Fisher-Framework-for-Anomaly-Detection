```python
import numpy as np
import pandas as pd
import os

output_dir = './summary_and_statistic_result'
```

# SUMMARY RESULTS FOR RQ1

```python
models_list = [ "KNN", "IForest", "OCSVM", "DeepSVDD", "PCA", "ECOD", "COPOD", "AutoEncoder", "LUNAR", "DIF", "XGBClassifier", "NeuTraLAD", "SFF_OD"] 

dir_path = './realworld_data_result'

def summarize_model_results(dir_path: str, models_list: list[str]):
    metric_cols = ["AUCROC", "AUCPR", "Accuracy", "MCC", 
                   "F1 Score", "Precision", "Recall", 
                   "Time Train", "Time Test"]

    all_results = []

    for model in models_list:
        file_name = f'{model}_result_realworld_data.csv'
        file_path = os.path.join(dir_path, file_name)

        if os.path.exists(file_path):
            df = pd.read_csv(file_path)

            if "Model" not in df.columns:
                df["Model"] = model

            all_results.append(df)
        else:
            print(f"Warning: {file_name} not found")

    # Gộp toàn bộ dữ liệu
    combined_df = pd.concat(all_results, ignore_index=True)

    print("Number of KFOLD in each model\n", combined_df.groupby("Model")['KFold'].agg(['count']))

    # Tính mean + std
    summary_df = combined_df.groupby("Model")[metric_cols].agg(['mean', 'std'])

    return summary_df, combined_df

summary, all_results = summarize_model_results(
    dir_path=dir_path, 
    models_list=models_list
)

# Lưu summary (flatten MultiIndex cho dễ đọc)
summary_flat = summary.copy()
summary_flat.columns = ['_'.join(col) for col in summary_flat.columns]

summary_flat.to_csv(os.path.join(output_dir, 'RQ1_realworld_statistic.csv'))

# Lưu toàn bộ raw results
all_results.to_csv(os.path.join(output_dir, 'RQ1_all_results.csv'), index=False)
```

```python
models_list = [ "KNN", "IForest", "OCSVM", "DeepSVDD", "PCA", "ECOD", "COPOD", "AutoEncoder", "LUNAR", "DIF", "XGBClassifier", "NeuTraLAD", "SFF_OD"] 

dir_path = './noise_realworld_data_result'

noise_types = ['add_duplicate_anomaly', 'irrelevant_features', 'label_contamination']

def summarize_model_noise_results(dir_path: str, models_list: list[str], noise_type: str):
    print("="*20 + f" {noise_type.upper()} " + "="*20)
    metric_cols = ["AUCROC", "AUCPR", "Accuracy", "MCC", 
                   "F1 Score", "Precision", "Recall", 
                   "Time Train", "Time Test"]

    all_results = []

    for model in models_list:
        file_name = f'{model}_{noise_type}_realworld_data.csv'
        file_path = os.path.join(dir_path, file_name)

        if os.path.exists(file_path):
            df = pd.read_csv(file_path)

            if "Model" not in df.columns:
                df["Model"] = model

            all_results.append(df)
        else:
            print(f"Warning: {file_name} not found")

    # Gộp toàn bộ dữ liệu
    combined_df = pd.concat(all_results, ignore_index=True)

    print("Number of KFOLD in each model\n", combined_df.groupby("Model")['KFold'].agg(['count']))

    # Tính mean + std
    summary_df = combined_df.groupby("Model")[metric_cols].agg(['mean', 'std'])

    return summary_df, combined_df

for noise_type in noise_types:
    summary, all_results = summarize_model_noise_results(
      dir_path=dir_path, 
      models_list=models_list,
      noise_type=noise_type
    )

    # Lưu summary (flatten MultiIndex cho dễ đọc)
    summary_flat = summary.copy()
    summary_flat.columns = ['_'.join(col) for col in summary_flat.columns]

    summary_flat.to_csv(os.path.join(output_dir, f'RQ2_{noise_type}_realworld_statistic.csv'))

    # Lưu toàn bộ raw results
    all_results.to_csv(os.path.join(output_dir, f'RQ2_{noise_type}_all_results.csv'), index=False)
```

# SUMMARY RESULTS FOR RQ3

```python
models_list = [ "KNN", "IForest", "OCSVM", "DeepSVDD", "PCA", "ECOD", "COPOD", "AutoEncoder", "LUNAR", "DIF", "XGBClassifier", "NeuTraLAD", "SFF_OD"] 

dir_path = './synthetic_data_result'

synthetic_types = ['cluster', 'dependency', 'global', 'local']

def summarize_model_noise_results(dir_path: str, models_list: list[str], synthetic_type: str):
    print("="*20 + f" {noise_type.upper()} " + "="*20)
    metric_cols = ["AUCROC", "AUCPR", "Accuracy", "MCC", 
                   "F1 Score", "Precision", "Recall", 
                   "Time Train", "Time Test"]

    all_results = []
    dir_path = os.path.join(dir_path, synthetic_type)

    for model in models_list:
        file_name = f'{model}_{synthetic_type}_synthetic_data.csv'
        file_path = os.path.join(dir_path, file_name)
 
        if os.path.exists(file_path):
            df = pd.read_csv(file_path)

            if "Model" not in df.columns:
                df["Model"] = model

            all_results.append(df)
        else:
            print(f"Warning: {file_name} not found")

    # Gộp toàn bộ dữ liệu
    combined_df = pd.concat(all_results, ignore_index=True)

    print("Number of DATASETS in each model\n", 
          combined_df.groupby("Model")['Dataset'].agg(['count']))

    for col in metric_cols:
        combined_df[col] = pd.to_numeric(combined_df[col], errors='coerce')
      
    combined_df[metric_cols] = combined_df[metric_cols].fillna(0)

    # Tính mean + std
    summary_df = combined_df.groupby("Model")[metric_cols].agg(['mean', 'std'])

    return summary_df, combined_df

for synthetic_type in synthetic_types:
    summary, all_results = summarize_model_noise_results(
      dir_path=dir_path, 
      models_list=models_list,
      synthetic_type=synthetic_type
    )

    # Lưu summary (flatten MultiIndex cho dễ đọc)
    summary_flat = summary.copy()
    summary_flat.columns = ['_'.join(col) for col in summary_flat.columns]

    summary_flat.to_csv(os.path.join(output_dir, f'RQ3_{synthetic_type}_realworld_statistic.csv'))

    # Lưu toàn bộ raw results
    all_results.to_csv(os.path.join(output_dir, f'RQ3_{synthetic_type}_all_results.csv'), index=False)
```
